```python
import re
import scanpy as sc
import anndata as ad
import pandas as pd

files = sorted(raw_dir.glob("kumar_*.fixed.h5ad"))
if not files:
    raise FileNotFoundError(f"No kumar_*.fixed.h5ad files in {raw_dir}")

adatas = []
sample_names = []

for f in files:
    gsm = re.search(r"GSM\d+", f.name).group()
    a = sc.read_h5ad(f)

    adatas.append(a)
    sample_names.append(gsm)

    print(gsm, a.shape)

assert len(sample_names) == len(set(sample_names)), "Duplicate GSM files"
```

```python
# 모든 파일의 gene ID와 순서가 동일한지 확인
assert all(
    a.var_names.equals(adatas[0].var_names)
    for a in adatas
), "Gene IDs or ordering differ between samples"

adata = ad.concat(
    adatas,
    keys=sample_names,
    label="GSM",
    index_unique="-",
    join="inner",
    merge="same",  # 동일한 gene_symbols, feature_types 유지
)

print(adata)
print(adata.obs["GSM"].value_counts())
```

```python
# GSM별 최종 metadata dictionary
metadata_by_gsm = {}

for gsm, values in gsm_rows.items():
    sample = dict(zip(gsm_columns, values))
    donor = sample["donor_id"]
    assay = sample["sample_id"]

    metadata_by_gsm[gsm] = {
        **common,
        **donor_metadata[donor],
        **sample,
        "donor_id_global": f"Kumar2022_{donor}",
        "donor_id_source": donor,
        "assay_id": assay,
        "tissue_id": f"Kumar2022_{assay}",
        "lobe": sample["brain_region"],
        "histology_notes": histology_notes[donor],
        "notes": (
            "Clinical Supplementary Table S5: "
            + clinical_notes[donor]
            + notes_suffix
        ),
    }

metadata = pd.DataFrame.from_dict(metadata_by_gsm, orient="index")
metadata.index.name = "GSM"

# 매핑되지 않는 GSM이 있으면 중단
gsm_per_cell = adata.obs["GSM"].astype(str)
unmapped = sorted(set(gsm_per_cell) - set(metadata.index))

if unmapped:
    raise ValueError(f"Metadata missing for GSMs: {unmapped}")

# GSM을 기준으로 모든 metadata column을 각 cell에 매핑
for column in metadata.columns:
    adata.obs[column] = gsm_per_cell.map(metadata[column])

# 확인
assert adata.obs["donor_id"].notna().all()
assert adata.obs["sample_id"].notna().all()

display(
    adata.obs[
        [
            "GSM", "donor_id", "sample_id",
            "diagnosis_raw", "fcd_subtype",
            "brain_region", "hemisphere",
        ]
    ]
    .drop_duplicates()
    .sort_values("GSM")
)

print(adata)
```

```python
# 숫자여야 하는 열: None → NaN
numeric_columns = [
    "age_years",
    "epilepsy_duration_years",
    "n_cells_raw",
    "n_cells_qc",
    "n_microglia",
]

for col in numeric_columns:
    if col in adata.obs:
        adata.obs[col] = pd.to_numeric(
            adata.obs[col], errors="raise"
        ).astype(float)

# 문자 metadata: None을 실제 결측값으로 유지하면서 저장 가능한 타입으로 변환
for col in adata.obs.select_dtypes(include=["object"]).columns:
    s = adata.obs[col]
    if s.dropna().map(lambda x: isinstance(x, str)).all():
        adata.obs[col] = pd.Categorical(s)
```
